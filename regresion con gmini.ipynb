```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

y=np.array([78.5,74.3,104.3,87.6,95.9,109.2,102.7,72.5,93.1,115.9,83.8,113.3,109.4])
x1=np.array([7,1,11,11,7,11,3,1,2,21,1,11,10])
x2=np.array([26,29,56,31,52,55,71,31,54,47,40,66,68])
x3=np.array([6,15,8,8,6,9,17,22,18,4,23,9,8])
x4=np.array([60,52,20,47,33,22,6,44,22,26,34,12,12])

dic={"y":y,"x1":x1,"x2":x2,"x3":x3,"x4":x4}
df=pd.DataFrame(dic)
df
```

```python
import numpy as np

# Extract the independent variables (x1, x2, x3, x4) from the DataFrame
X = df[['x1', 'x2', 'x3', 'x4']].values

# Create a column of ones with the same number of rows as X
ones = np.ones((X.shape[0], 1))

# Concatenate the column of ones with the X matrix
X_with_intercept = np.concatenate((ones, X), axis=1)

print("Original X matrix (first 5 rows):")
print(X[:5])
print("\nMatrix with first column replaced by ones (first 5 rows):")
print(X_with_intercept[:5])
```

```python
# Extract the 'y' column from the DataFrame
y_column = df['y'].values

# Reshape the 1D array into a column vector
y_vector = y_column.reshape(-1, 1)

print("Original 'y' values (first 5 elements):")
print(y_column[:5])
print("\n'y' column as a vector (first 5 elements and shape):")
print(y_vector[:5])
print(f"Shape of y_vector: {y_vector.shape}")
```

```python
# Calculate the transpose of X_with_intercept
X_with_intercept_T = X_with_intercept.T

print("Transpose of X_with_intercept matrix:")
print(X_with_intercept_T)
```

```python
# Perform matrix multiplication: X_with_intercept_T * X_with_intercept
product_XT_X = np.dot(X_with_intercept_T, X_with_intercept)

print("Result of (X_with_intercept_T * X_with_intercept):")
print(product_XT_X)
print(f"Shape of the resulting matrix: {product_XT_X.shape}")
```

```python
# Calculate the inverse of product_XT_X
inverse_product_XT_X = np.linalg.inv(product_XT_X)

print("Inverse of (X_with_intercept_T * X_with_intercept):")
print(inverse_product_XT_X)
print(f"Shape of the inverse matrix: {inverse_product_XT_X.shape}")
```
